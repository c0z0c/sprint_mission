```python
# ============================================================================
# 4. OpenVINO INT8 양자화
# ============================================================================
from openvino.runtime import Core
import glob

logger.info("[3/3] OpenVINO INT8 모델 내보내기 중...")

# 기존 디렉토리 강제 삭제
output_openvino_int8_dir = my_driver_path(yolov8m_path, 'weights', 'best_int8_openvino_model', create=True)
if os.path.exists(output_openvino_int8_dir):
    logger.info(f"기존 OpenVINO 디렉토리 삭제 중: {output_openvino_int8_dir}")
    try:
        # 읽기 전용 파일도 삭제할 수 있도록 권한 변경
        import stat
        def remove_readonly(func, path, _):
            os.chmod(path, stat.S_IWRITE)
            func(path)
        
        shutil.rmtree(output_openvino_int8_dir, onerror=remove_readonly)
        logger.info("  ✓ 삭제 완료")
    except Exception as e:
        logger.warning(f"디렉토리 삭제 실패: {e}")

# OpenVINO INT8 export
logger.info("OpenVINO INT8 양자화 시작...")
output_openvino_int8_path = model.export(
    format='openvino',
    int8=True,      # INT8 양자화
    imgsz=640,
    data=yaml_path,
)

logger.info(f"OpenVINO INT8 내보내기 완료: {output_openvino_int8_path}")

# 결과 확인
if os.path.exists(output_openvino_int8_path):
    openvino_files = os.listdir(output_openvino_int8_path) if os.path.isdir(output_openvino_int8_path) else [output_openvino_int8_path]
    logger.info(f"생성된 파일: {openvino_files}")
    
    # 크기 계산
    if os.path.isdir(output_openvino_int8_path):
        total_size = sum(os.path.getsize(os.path.join(output_openvino_int8_path, f)) 
                        for f in openvino_files if os.path.isfile(os.path.join(output_openvino_int8_path, f)))
    else:
        total_size = os.path.getsize(output_openvino_int8_path)
    
    openvino_int8_size = total_size / (1024 * 1024)
    logger.info(f"OpenVINO INT8 모델 크기: {openvino_int8_size:.2f} MB")
    
    # 모델 로드 - 실제 생성된 XML 파일 찾기
    xml_files = glob.glob(os.path.join(output_openvino_int8_path, "*.xml"))
    if xml_files:
        xml_path = xml_files[0]
        logger.info(f"발견된 XML 파일: {os.path.basename(xml_path)}")
        
        core = Core()
        openvino_model = core.read_model(model=xml_path)
        
        # 구조 출력
        with open(output_openvino_int8_path + "/openvino_structure.txt", "w") as f:
            for op in openvino_model.get_ops():
                output_shapes = [o.get_shape() for o in op.outputs()]  # outputs() 호출 필요
                f.write(f"{op.get_friendly_name()} ({op.get_type_name()}) -> outputs: {output_shapes}\n")                
        logger.info("  ✓ OpenVINO 모델 구조 저장 완료")
    else:
        logger.warning("  ⚠ XML 파일을 찾을 수 없습니다")
```

```python
# ============================================================================
# 5. ONNX FP32 → INT8 QDQ 양자화 준비
# ============================================================================
logger.info("="*80)
logger.info("ONNX Runtime Static Quantization (INT8 - QDQ 형식)")
logger.info("="*80)
logger.info("ONNX FP32 모델을 INT8 QDQ 형식으로 양자화합니다.")
logger.info("QDQ (Quantize-Dequantize): CPU에서 실행 가능한 형식")
logger.info("실제 데이터셋 이미지로 Calibration 수행 (랜덤 노이즈 X)")
logger.info("")

# 1. ONNX Runtime 설치 확인
logger.info("Step 1: ONNX Runtime 설치 확인")
install_if_missing("onnxruntime")

from onnxruntime.quantization import quantize_static, QuantFormat, QuantType
from onnxruntime.quantization.calibrate import CalibrationDataReader
from onnxruntime.transformers.onnx_model import OnnxModel
import onnx
import numpy as np
import cv2

logger.info("  ✓ ONNX Runtime 로드 완료")

# 2. 데이터셋 재생성 (1000개씩)
logger.info("\nStep 2: 데이터셋 재생성 (Calibration용)")
logger.setLevel(logging.INFO)
yolo_dataset_path = my_cache_path("yolo", "the-oxfordiiit-pet-dataset")
yaml_path, train_df, valid_df, test_df, validation_results = oxfordiit_pet_to_yolo(
    max_samples_per_split=(1000, 1000, 1000),
    label_mode="species",
    output_dir=Path(yolo_dataset_path)
)
logger.setLevel(logging.DEBUG)

logger.info(f"  ✓ 데이터셋 경로: {yolo_dataset_path}")
logger.info(f"  ✓ Train: {len(train_df)}개, Valid: {len(valid_df)}개, Test: {len(test_df)}개")


# 3. 실제 데이터 기반 Calibration Reader 정의
class RealDataCalibrationReader(CalibrationDataReader):
    """실제 데이터셋 이미지로 Calibration 수행 (YOLOv8 표준 전처리 적용)"""
    
    def __init__(self, dataset_path, num_samples=300):
        self.dataset_path = dataset_path
        self.num_samples = num_samples
        self.image_files = self._load_image_list()
        self.current = 0
        logger.info(f"  ✓ RealDataCalibrationReader 초기화: {len(self.image_files)}개 이미지 로드")
    
    def _load_image_list(self):
        """train + valid 이미지 수집 (test는 제외)"""
        images = []
        for split in ['train', 'valid']:
            split_path = os.path.join(self.dataset_path, 'images', split)
            if os.path.exists(split_path):
                imgs = [f for f in os.listdir(split_path)
                       if f.lower().endswith(('.jpg', '.jpeg', '.png'))]
                images.extend([os.path.join(split_path, img) for img in imgs])
        
        # num_samples만큼 샘플링
        if len(images) > self.num_samples:
            images = images[:self.num_samples]
        
        logger.info(f"  ✓ Calibration 이미지 수집: {len(images)}개")
        return images
    
    def _preprocess_image(self, img_path):
        """YOLOv8 표준 전처리 (LetterBox + BGR2RGB + 정규화)"""
        try:
            from ultralytics.data.augment import LetterBox
            
            # 이미지 로드
            img = cv2.imread(img_path)
            if img is None:
                raise ValueError(f"이미지 로드 실패: {img_path}")
            
            # LetterBox 리사이즈 (auto=False로 정확히 640x640 보장)
            letterbox = LetterBox(640, auto=False, stride=32)
            img = letterbox(image=img)
            
            # BGR → RGB
            img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
            
            # 정규화 [0, 1]
            img = img.astype(np.float32) / 255.0
            
            # HWC → CHW
            img = img.transpose(2, 0, 1)
            
            # 배치 차원 추가
            img_batch = img[np.newaxis, :]
            
            # 디버깅: 처음 3개 이미지만 크기 확인
            if self.current < 3:
                logger.info(f"  [이미지 {self.current}] {os.path.basename(img_path)}: 최종 크기={img_batch.shape}")
            
            # 크기 검증
            if img_batch.shape != (1, 3, 640, 640):
                raise ValueError(f"전처리 후 크기 불일치: {img_batch.shape} != (1, 3, 640, 640)")
            
            return img_batch
            
        except Exception as e:
            logger.error(f"이미지 전처리 실패 ({img_path}): {e}")
            raise
    
    def get_next(self):
        """다음 calibration 데이터 반환"""
        if self.current >= len(self.image_files):
            return None
        
        img_path = self.image_files[self.current]
        self.current += 1
        
        # 진행 상황 표시 (10%마다)
        if self.current % max(1, len(self.image_files) // 10) == 0:
            progress = (self.current / len(self.image_files)) * 100
            logger.info(f"  Calibration 진행: {self.current}/{len(self.image_files)} ({progress:.0f}%)")
        
        try:
            img_tensor = self._preprocess_image(img_path)
            return {'images': img_tensor}
        except Exception as e:
            logger.warning(f"이미지 로드 실패 ({img_path}): {e}")
            # 다음 이미지 시도
            return self.get_next()
    
    def rewind(self):
        """리더 초기화"""
        self.current = 0


# 4. 입력 ONNX FP32 모델 경로 확인
logger.info("\nStep 3: ONNX FP32 모델 경로 확인")

if 'onnx_fp32_final' in dir() and os.path.exists(onnx_fp32_final):
    onnx_fp32_path = onnx_fp32_final
elif 'onnx_fp32_export' in dir() and os.path.exists(onnx_fp32_export):
    onnx_fp32_path = onnx_fp32_export
else:
    onnx_fp32_path = os.path.join(yolov8m_path, 'weights', 'best.onnx')

if not os.path.exists(onnx_fp32_path):
    raise FileNotFoundError(f"ONNX FP32 모델을 찾을 수 없습니다: {onnx_fp32_path}")

fp32_size = os.path.getsize(onnx_fp32_path) / (1024 * 1024)
logger.info(f"  ✓ ONNX FP32 모델 크기: {fp32_size:.2f} MB")
logger.info(f"  ✓ 모델 경로: {onnx_fp32_path}")


# 5. 출력 경로 설정
logger.info("\nStep 4: 출력 경로 설정")
output_onnx_int8_path = my_driver_path(yolov8m_path, 'weights', 'out_onnx_int8_qdq', create=True)
onnx_int8_final = os.path.join(output_onnx_int8_path, 'yolov8m_int8_qdq.onnx')
logger.info(f"  ✓ INT8 QDQ 모델 저장 경로: {onnx_int8_final}")


# 6. Calibration Reader 생성
logger.info("\nStep 5: Calibration Reader 생성")
calibration_reader = RealDataCalibrationReader(
    dataset_path=yolo_dataset_path,
    num_samples=500  # 300-500개 권장
)
```

```python
# ============================================================================
# 6-1. 출력 노드 자동 탐지 (양자화 제외 대상) - 최종 수정 버전
# ============================================================================
logger.info("Step 6-1: 출력 노드 자동 탐지 (양자화 제외)")
logger.info("  - YOLO 출력 레이어(로짓)는 양자화하면 안 됩니다")
logger.info("  - 출력 노드부터 역추적하여 모든 출력 관련 노드 제외")
logger.info("  - Div, Sub 등 산술 연산도 포함하여 추적")
logger.info("")

import onnx

# FP32 모델 로드
fp32_model = onnx.load(onnx_fp32_final)
output_node_name = fp32_model.graph.output[0].name
logger.info(f"  ✓ 출력 노드: {output_node_name}")

# 출력에 연결된 노드 찾기 (양자화 제외 대상)
nodes_to_exclude = []

def find_parent_nodes(node_name, depth=0, max_depth=5):
    """재귀적으로 부모 노드 찾기"""
    if depth >= max_depth:
        return
    
    for node in fp32_model.graph.node:
        if node_name in node.output:
            if node.name not in nodes_to_exclude:
                nodes_to_exclude.append(node.name)
                logger.info(f"  [깊이 {depth}] 제외: {node.name[:60]} (타입: {node.op_type})")
                
                # 산술/결합 연산이면 모든 입력 추적
                # Concat, Mul, Add, Div, Sub, Transpose, Reshape 등
                if node.op_type in ['Concat', 'Mul', 'Add', 'Div', 'Sub', 
                                   'Transpose', 'Reshape', 'Sigmoid']:
                    for input_name in node.input:
                        find_parent_nodes(input_name, depth + 1, max_depth)
            break

# 출력 노드부터 역추적 (깊이 5단계)
find_parent_nodes(output_node_name, depth=0, max_depth=5)

# 중복 제거
nodes_to_exclude = list(set(nodes_to_exclude))

logger.info(f"\n  ✓ 양자화 제외 노드 총 {len(nodes_to_exclude)}개:")
for idx, node_name in enumerate(nodes_to_exclude):
    logger.info(f"    [{idx+1}] {node_name}")
logger.info("")
```

```python
# ============================================================================
# 6-2. Static Quantization with QDQ format (op_types 제한 방식)
# ============================================================================
from onnxruntime.quantization import quantize_static, QuantFormat, QuantType

logger.info("\n" + "="*80)
logger.info("Step 6-2: Static Quantization (QDQ 형식) - op_types 제한 방식")
logger.info("="*80)
logger.info("  - Concat 노드 제외 (출력 레이어 보호)")
logger.info("  - Conv, MatMul 등 주요 연산만 양자화")
logger.info("  - Per-channel quantization (정밀도 향상)")
logger.info("  - 처리 중... (약 2-5분 소요)")
logger.info("")

# 출력 경로 (기존 경로 유지)
output_onnx_int8_path = my_driver_path(yolov8m_path, 'weights', 'out_onnx_int8_qdq', create=True)
onnx_int8_final = os.path.join(output_onnx_int8_path, 'yolov8m_int8_qdq.onnx')

# Calibration Reader 초기화
calibration_reader.rewind()

# quantize_static(
#     model_input=onnx_fp32_final,
#     model_output=onnx_int8_final,
#     calibration_data_reader=calibration_reader,
#     quant_format=QuantFormat.QDQ,
#     activation_type=QuantType.QInt8,
#     weight_type=QuantType.QInt8,
#     # 핵심: 양자화할 레이어 타입만 명시 (Concat 제외)
#     op_types_to_quantize=[
#         'Conv',      # 합성곱 레이어
#         'MatMul',    # 행렬 곱셈
#         'Gemm',      # 일반 행렬 곱셈
#         'Add',       # 덧셈 (Residual connection)
#         'Mul',       # 곱셈
#     ],
#     per_channel=True,
#     reduce_range=False,
#     extra_options={
#         'ActivationSymmetric': False,
#         'WeightSymmetric': True,
#     }
# )

quantize_static(
    model_input=onnx_fp32_final,
    model_output=onnx_int8_final,
    calibration_data_reader=calibration_reader,
    quant_format=QuantFormat.QDQ,
    activation_type=QuantType.QInt8,
    weight_type=QuantType.QInt8,
    nodes_to_exclude=nodes_to_exclude,  # Cell #VSC-a6ead63e에서 추출한 노드 리스트
    per_channel=True,
    reduce_range=False,
    extra_options={
        'ActivationSymmetric': False,
        'WeightSymmetric': True,
    }
)

logger.info("\n  ✓ Static Quantization (QDQ) 완료 (op_types 제한)")
logger.info("")

# 검증
onnx_model = onnx.load(onnx_int8_final)
onnx.checker.check_model(onnx_model)

# 출력 직전 노드 확인
output_node_name = onnx_model.graph.output[0].name
for node in onnx_model.graph.node:
    if output_node_name in node.output:
        logger.info(f"출력 직전 노드: {node.name} (타입: {node.op_type})")
        if node.op_type == 'DequantizeLinear':
            logger.error("  여전히 양자화되어 있습니다!")
        else:
            logger.info("  양자화 제외됨 (정상)")
        break

# 구조 저장
with open(onnx_int8_final + ".txt", "w") as f:
    f.write("=== ONNX Graph Structure (INT8 QDQ) ===\n")
    f.write(str(onnx.helper.printable_graph(onnx_model.graph)))
    f.write("\n\n=== ONNX Model Info (INT8 QDQ) ===\n")
    f.write(f"IR Version: {onnx_model.ir_version}\n")
    f.write(f"Producer: {onnx_model.producer_name} {onnx_model.producer_version}\n")
    f.write(f"Total nodes: {len(onnx_model.graph.node)}\n")
    
    f.write("\nInputs:\n")
    for i in onnx_model.graph.input:
        f.write(f"  {i.name} : {i.type}\n")
    
    f.write("\nOutputs:\n")
    for o in onnx_model.graph.output:
        f.write(f"  {o.name} : {o.type}\n")
    
    f.write("\n=== 출력 노드 양자화 검증 ===\n")
    for node in onnx_model.graph.node:
        if output_node_name in node.output:
            f.write(f"출력 직전 노드: {node.name} (타입: {node.op_type})\n")
            if node.op_type == 'DequantizeLinear':
                f.write("경고: 출력 노드가 여전히 양자화되어 있습니다!\n")
            else:
                f.write("출력 노드 양자화 제외됨 (정상)\n")

# 크기 확인
int8_size = os.path.getsize(onnx_int8_final) / (1024 * 1024)
compression_ratio = (1 - int8_size / fp32_size) * 100

logger.info("\n" + "="*80)
logger.info("양자화 결과")
logger.info("="*80)
logger.info(f"  - FP32 모델 크기: {fp32_size:.2f} MB")
logger.info(f"  - INT8 QDQ 모델 크기: {int8_size:.2f} MB")
logger.info(f"  - 압축률: {compression_ratio:.1f}%")
logger.info(f"  - Calibration 방식: 실제 데이터셋 이미지 ({len(calibration_reader.image_files)}개)")
logger.info(f"  - 양자화 대상: Conv, MatMul, Gemm, Add, Mul")
logger.info(f"  - 제외: Concat, Reshape, Transpose 등")
logger.info(f"  - Per-channel quantization: 활성화")
logger.info("")
logger.info(f"  - FP32 경로: {onnx_fp32_path}")
logger.info(f"  - INT8 QDQ 경로: {onnx_int8_final}")
logger.info("="*80)
```

```python
# ============================================================================
# ONNX 모델 구조 분석 및 저장 (FP32 vs INT8 QDQ 비교)
# ============================================================================
logger.info("="*80)
logger.info("ONNX 모델 구조 분석 및 저장")
logger.info("="*80)

import onnx

def save_onnx_model_info(model_path, output_txt_path):
    """ONNX 모델 구조를 텍스트 파일로 저장"""
    try:
        model = onnx.load(model_path)
        
        with open(output_txt_path, 'w', encoding='utf-8') as f:
            f.write("="*80 + "")
            f.write(f"ONNX 모델 구조: {os.path.basename(model_path)}\n")
            f.write("="*80 + "")
            
            # 1. 기본 정보
            f.write("1. 기본 정보\n")
            f.write("-"*80 + "")
            f.write(f"IR 버전: {model.ir_version}\n")
            f.write(f"Producer: {model.producer_name} {model.producer_version}\n")
            f.write(f"ONNX 버전: {model.opset_import[0].version}\n")
            f.write(f"그래프 이름: {model.graph.name}\n")
            f.write("")
            
            # 2. 입력 정보
            f.write("2. 입력 정보\n")
            f.write("-"*80 + "")
            for idx, input_node in enumerate(model.graph.input):
                f.write(f"입력 {idx}: {input_node.name}\n")
                if input_node.type.tensor_type.HasField('shape'):
                    shape = [dim.dim_value if dim.HasField('dim_value') else 'dynamic' 
                            for dim in input_node.type.tensor_type.shape.dim]
                    f.write(f"  Shape: {shape}\n")
                f.write(f"  Type: {onnx.TensorProto.DataType.Name(input_node.type.tensor_type.elem_type)}\n")
            f.write("")
            
            # 3. 출력 정보
            f.write("3. 출력 정보\n")
            f.write("-"*80 + "")
            for idx, output_node in enumerate(model.graph.output):
                f.write(f"출력 {idx}: {output_node.name}\n")
                if output_node.type.tensor_type.HasField('shape'):
                    shape = [dim.dim_value if dim.HasField('dim_value') else 'dynamic' 
                            for dim in output_node.type.tensor_type.shape.dim]
                    f.write(f"  Shape: {shape}\n")
                f.write(f"  Type: {onnx.TensorProto.DataType.Name(output_node.type.tensor_type.elem_type)}\n")
            f.write("")
            
            # 4. 노드 통계
            f.write("4. 노드 통계\n")
            f.write("-"*80 + "")
            f.write(f"전체 노드 수: {len(model.graph.node)}\n")
            
            # 노드 타입별 카운트
            node_types = {}
            for node in model.graph.node:
                node_types[node.op_type] = node_types.get(node.op_type, 0) + 1
            
            f.write(f"노드 타입 종류: {len(node_types)}개\n")
            f.write("노드 타입별 개수:\n")
            for op_type, count in sorted(node_types.items(), key=lambda x: x[1], reverse=True):
                f.write(f"  {op_type}: {count}개\n")
            f.write("")
            
            # 5. QDQ 노드 확인 (INT8 모델만)
            qdq_nodes = [n for n in model.graph.node if 'Quantize' in n.op_type or 'Dequantize' in n.op_type]
            if qdq_nodes:
                f.write("5. QDQ (Quantize/Dequantize) 노드 정보\n")
                f.write("-"*80 + "")
                f.write(f"QDQ 노드 총 개수: {len(qdq_nodes)}\n")
                
                quantize_nodes = [n for n in qdq_nodes if n.op_type == 'QuantizeLinear']
                dequantize_nodes = [n for n in qdq_nodes if n.op_type == 'DequantizeLinear']
                
                f.write(f"  - QuantizeLinear: {len(quantize_nodes)}개\n")
                f.write(f"  - DequantizeLinear: {len(dequantize_nodes)}개\n")
                f.write("")
                
                # 첫 번째 QDQ 노드 상세 정보
                if len(qdq_nodes) > 0:
                    f.write("첫 번째 QDQ 노드 예시:\n")
                    first_qdq = qdq_nodes[0]
                    f.write(f"  이름: {first_qdq.name}\n")
                    f.write(f"  타입: {first_qdq.op_type}\n")
                    f.write(f"  입력: {list(first_qdq.input)}\n")
                    f.write(f"  출력: {list(first_qdq.output)}\n")
                f.write("")
            
            # 6. 상세 노드 리스트 (처음 20개만)
            f.write("6. 상세 노드 리스트 (처음 20개)\n")
            f.write("-"*80 + "")
            for idx, node in enumerate(model.graph.node[:20]):
                f.write(f"노드 {idx}: {node.name if node.name else f'unnamed_{idx}'}\n")
                f.write(f"  타입: {node.op_type}\n")
                f.write(f"  입력: {list(node.input)}\n")
                f.write(f"  출력: {list(node.output)}\n")
                if node.attribute:
                    f.write(f"  속성: {len(node.attribute)}개\n")
            
            if len(model.graph.node) > 20:
                f.write(f"... (나머지 {len(model.graph.node) - 20}개 노드 생략)\n")
            
            f.write("")
            f.write("="*80 + "")
            f.write("구조 분석 완료\n")
            f.write("="*80 + "")
        
        return True
    except Exception as e:
        logger.error(f"모델 구조 저장 실패 ({model_path}): {e}")
        return False


# FP32 모델 구조 저장
fp32_structure_path = os.path.join(output_onnx_fp32_path, 'yolov8m_fp32_structure.txt')
logger.info(f"FP32 모델 구조 저장 중: {fp32_structure_path}")
if save_onnx_model_info(onnx_fp32_path, fp32_structure_path):
    logger.info(f"  ✓ 저장 완료")
else:
    logger.error(f"  ✗ 저장 실패")

# INT8 QDQ 모델 구조 저장
int8_structure_path = os.path.join(output_onnx_int8_path, 'yolov8m_int8_qdq_structure.txt')
logger.info(f"INT8 QDQ 모델 구조 저장 중: {int8_structure_path}")
if save_onnx_model_info(onnx_int8_final, int8_structure_path):
    logger.info(f"  ✓ 저장 완료")
else:
    logger.error(f"  ✗ 저장 실패")

logger.info("")
logger.info("="*80)
logger.info("모델 구조 비교")
logger.info("="*80)
logger.info(f"FP32 구조: {fp32_structure_path}")
logger.info(f"INT8 구조: {int8_structure_path}")
logger.info("")
logger.info("비교 방법:")
logger.info("  1. 두 파일을 텍스트 에디터로 열기")
logger.info("  2. '4. 노드 통계' 섹션에서 QDQ 노드 개수 확인")
logger.info("  3. 'QuantizeLinear', 'DequantizeLinear' 노드가 INT8 모델에만 있는지 확인")
logger.info("  4. 출력 노드 직전에 DequantizeLinear가 있으면 → 출력이 양자화됨 (문제 가능성)")
logger.info("="*80)
```

```python
# ============================================================================
# ONNX INT8 QDQ 모델 수동 검증 및 출력 분석
# ============================================================================
logger.info("="*80)
logger.info("ONNX INT8 QDQ 모델 출력 검증 (수정 후)")
logger.info("="*80)

import onnxruntime as ort

# ONNX Runtime 세션 생성
session = ort.InferenceSession(onnx_int8_final)
input_name = session.get_inputs()[0].name
input_shape = session.get_inputs()[0].shape
output_names = [output.name for output in session.get_outputs()]

logger.info(f"입력 이름: {input_name}")
logger.info(f"입력 shape: {input_shape}")
logger.info(f"출력 이름: {output_names}")
logger.info("")

# 랜덤 입력으로 추론 테스트
logger.info("Step 1: 랜덤 입력으로 추론 테스트")
dummy_input = np.random.rand(1, 3, 640, 640).astype(np.float32)
logger.info(f"  - 입력 shape: {dummy_input.shape}")
logger.info(f"  - 입력 dtype: {dummy_input.dtype}")
logger.info(f"  - 입력 범위: [{dummy_input.min():.3f}, {dummy_input.max():.3f}]")

outputs = session.run(None, {input_name: dummy_input})

logger.info("")
logger.info("Step 2: 출력 분석")
logger.info(f"  - 출력 shape: {outputs[0].shape}")
logger.info(f"  - 출력 dtype: {outputs[0].dtype}")
logger.info(f"  - 출력 범위: [{outputs[0].min():.6f}, {outputs[0].max():.6f}]")
logger.info("")

# YOLOv8 출력 형식: (batch, num_predictions, 5+nc)
# 5 = [x, y, w, h, objectness_logit] (로짓 형태, sigmoid 미적용)
if len(outputs[0].shape) == 3:
    batch, num_preds, features = outputs[0].shape
    logger.info(f"YOLOv8 출력 형식 확인:")
    logger.info(f"  - Batch 크기: {batch}")
    logger.info(f"  - 예측 박스 수: {num_preds}")
    logger.info(f"  - Feature 차원: {features} (예상: {5 + 2} = 7, x/y/w/h/objectness_logit + 2 class_logits)")
    logger.info("")
    
    # Objectness Logit 분석 (5번째 열) - 로짓이므로 -10~10 범위 기대
    objectness_logit = outputs[0][:, :, 4]
    logger.info(f"Step 3: Objectness Logit 분석 (4번째 인덱스)")
    logger.info(f"  - Logit 최소값: {objectness_logit.min():.6f}")
    logger.info(f"  - Logit 최대값: {objectness_logit.max():.6f}")
    logger.info(f"  - Logit 평균값: {objectness_logit.mean():.6f}")
    logger.info(f"  - 예상 범위: -10 ~ 10 (로짓)")
    
    # Sigmoid 적용 후 confidence 계산
    import scipy.special
    confidence = scipy.special.expit(objectness_logit)  # sigmoid
    logger.info("")
    logger.info(f"Step 3-1: Sigmoid 적용 후 Confidence 분석")
    logger.info(f"  - Confidence 최소값: {confidence.min():.6f}")
    logger.info(f"  - Confidence 최대값: {confidence.max():.6f}")
    logger.info(f"  - Confidence 평균값: {confidence.mean():.6f}")
    logger.info(f"  - Confidence > 0.001 개수: {(confidence > 0.001).sum()} / {confidence.size}")
    logger.info(f"  - Confidence > 0.001 비율: {(confidence > 0.001).sum() / confidence.size * 100:.2f}%")
    logger.info(f"  - Confidence > 0.25 개수: {(confidence > 0.25).sum()} / {confidence.size}")
    logger.info("")
    
    # 클래스 로짓 분석 (6번째 열 이후)
    if features > 5:
        class_logits = outputs[0][:, :, 5:]
        logger.info(f"Step 4: 클래스 로짓 분석 (5번째 인덱스 이후)")
        logger.info(f"  - 클래스 로짓 shape: {class_logits.shape}")
        logger.info(f"  - 클래스 로짓 범위: [{class_logits.min():.6f}, {class_logits.max():.6f}]")
        logger.info(f"  - 클래스 로짓 평균: {class_logits.mean():.6f}")
        logger.info(f"  - 예상 범위: -10 ~ 10 (로짓)")

logger.info("")
logger.info("="*80)
logger.info("검증 완료")
logger.info("="*80)
logger.info("")
logger.info("수정 후 기대 결과:")
logger.info("  1. Logit 범위: -10 ~ 10 (정상적인 로짓 범위)")
logger.info("  2. Sigmoid 적용 후 Confidence: 0 ~ 1 범위")
logger.info("  3. Confidence > 0.25 개수 > 0 (NMS 통과 예상)")
logger.info("")
logger.info("이전 문제 (수정 전):")
logger.info("  1. 출력 범위: 0 ~ 644 (비정상)")
logger.info("  2. Confidence 최대값: 108 (비정상)")
logger.info("  3. Confidence > 0.25 개수: 4 (모두 필터링됨)")
logger.info("")
```

```python
# ============================================================================
# 출력 노드 양자화 상태 정밀 진단
# ============================================================================
import onnx

logger.info("="*80)
logger.info("INT8 QDQ 모델 출력 레이어 정밀 진단")
logger.info("="*80)

# 1. nodes_to_exclude 변수 확인
logger.info(f"\n1. nodes_to_exclude 변수 검증")
logger.info(f"  - 타입: {type(nodes_to_exclude)}")
logger.info(f"  - 길이: {len(nodes_to_exclude)}")
logger.info(f"  - 내용: {nodes_to_exclude}")

# 2. INT8 모델 로드 및 출력 노드 역추적
int8_model = onnx.load(onnx_int8_final)
output_name = int8_model.graph.output[0].name

logger.info(f"\n2. 출력 노드 역추적 (INT8 모델)")
logger.info(f"  - 출력 이름: {output_name}")

# 출력 직전 노드 찾기 (최대 5단계)
current_name = output_name
for depth in range(5):
    found = False
    for node in int8_model.graph.node:
        if current_name in node.output:
            logger.info(f"  [{depth}단계] 노드: {node.name[:50]} (타입: {node.op_type})")
            if node.op_type == 'DequantizeLinear':
                logger.error(f"    출력 경로에 DequantizeLinear 발견!")
            elif node.op_type == 'Concat':
                logger.info(f"    ✓ Concat 노드 발견")
                # Concat 입력 확인
                logger.info(f"    → 입력 개수: {len(node.input)}")
                for i, inp in enumerate(node.input[:3]):  # 처음 3개만
                    logger.info(f"      입력[{i}]: {inp[:50]}")
            
            # 다음 단계 준비
            if node.input:
                current_name = node.input[0]
                found = True
                break
    
    if not found:
        logger.info(f"  [{depth}단계] 더 이상 부모 노드 없음")
        break

# 3. FP32 모델과 비교
logger.info(f"\n3. FP32 모델과 비교")
fp32_model = onnx.load(onnx_fp32_final)
fp32_output_name = fp32_model.graph.output[0].name
logger.info(f"  - FP32 출력 이름: {fp32_output_name}")

for node in fp32_model.graph.node:
    if fp32_output_name in node.output:
        logger.info(f"  - FP32 출력 직전 노드: {node.name[:50]} (타입: {node.op_type})")
        break

logger.info("\n" + "="*80)
logger.info("진단 완료")
logger.info("="*80)
```
