```python
# 1. 설정 정보 (사용자님이 제공해주신 토큰 사용)
BASE_URL = "https://api.themoviedb.org/3"

# 2. 헤더 설정 (인증 토큰 포함)
headers = {
    "accept": "application/json",
    "Authorization": f"Bearer {ACCESS_TOKEN}"
}

def get_popular_movies():
    # 인기 영화 목록 엔드포인트
    url = f"{BASE_URL}/movie/popular?language=ko-KR&page=1"
    
    response = requests.get(url, headers=headers)
    
    if response.status_code == 200:
        data = response.json()
        movies = data.get('results', [])
        
        print(f"--- 총 {len(movies)}개의 영화를 찾았습니다 ---\n")
        for movie in movies:
            title = movie['title']
            release_date = movie.get('release_date', '정보 없음')
            vote_average = movie.get('vote_average', 0)
            
            print(f"🎬 제목: {title}")
            print(f"📅 개봉: {release_date}")
            print(f"⭐ 평점: {vote_average}")
            print("-" * 30)
    else:
        print(f"오류 발생: {response.status_code}")
        print(response.text)

# 실행
get_popular_movies()
```

```python
BASE_URL = "https://api.themoviedb.org/3"
headers = {
    "accept": "application/json",
    "Authorization": f"Bearer {ACCESS_TOKEN}"
}

def get_korean_movies_last_5_years(max_pages=10):
    all_movies = []
    
    # 파라미터 설명:
    # language=ko-KR: 제목과 줄거리를 한글로 가져옴
    # region=KR: 한국 개봉일 기준 적용
    # sort_by=primary_release_date.desc: 최신 개봉일 순 정렬
    # primary_release_date.gte=2021-01-01: 2021년부터
    # primary_release_date.lte=2025-12-31: 2025년까지
    
    params = {
        "language": "ko-KR",
        "region": "KR",
        "sort_by": "primary_release_date.desc",
        "primary_release_date.gte": "2021-01-01",
        "primary_release_date.lte": "2025-12-31",
        "include_adult": "false"
    }

    print("--- 최근 5개년 한글 데이터 수집 시작 ---")
    
    for page in range(1, max_pages + 1):
        params["page"] = page
        response = requests.get(f"{BASE_URL}/discover/movie", headers=headers, params=params)
        
        if response.status_code == 200:
            data = response.json()
            movies = data.get('results', [])
            
            if not movies:
                break
                
            all_movies.extend(movies)
            print(f"Page {page} 수집 완료 (누적 {len(all_movies)}건)")
        else:
            print(f"오류 발생: {response.status_code}")
            break
            
    return pd.DataFrame(all_movies)

# 2. 실행 및 필터링
df_raw = get_korean_movies_last_5_years(max_pages=15)

# 분석에 필요한 주요 컬럼만 추출
selected_cols = ['id', 'title', 'release_date', 'vote_average', 'popularity', 'overview']
df_korean = df_raw[selected_cols]

# 3. 데이터 전처리 (개봉일이 없는 데이터 제거)
df_korean = df_korean.dropna(subset=['release_date'])

print("\n--- 수집 완료된 데이터 샘플 (최신순) ---")
display(df_korean.head(100))
```

```python
import requests
import pandas as pd
import time

def get_tmdb_movies_by_period(start_date, end_date, max_pages=5):
    """
    특정 기간 동안의 영화 목록을 가져와 DataFrame으로 반환합니다.
    
    :param start_date: 시작일 (YYYY-MM-DD)
    :param end_date: 종료일 (YYYY-MM-DD)
    :param max_pages: 수집할 최대 페이지 수 (페이지당 20건)
    """
    # 설정 정보 (제공해주신 토큰 사용)
    ACCESS_TOKEN = "eyJhbGciOiJIUzI1NiJ9.eyJhdWQiOiIzNTdmZDE5NjM0YjU5ZTZhZjRlYTViN2FlZDAwODhmYiIsIm5iZiI6MTc1ODY3MTQzOS43NCwic3ViIjoiNjhkMzMyNGY0NDI2ZDNlOGE1MmVjMjI1Iiwic2NvcGVzIjpbImFwaV9yZWFkIl0sInZlcnNpb24iOjF9.Y0v4otYycGeGRt-ev-7LIHZLjKb72kLxGfDTlIFTz2w"
    BASE_URL = "https://api.themoviedb.org/3/discover/movie"
    
    headers = {
        "accept": "application/json",
        "Authorization": f"Bearer {ACCESS_TOKEN}"
    }
    
    all_movies = []
    
    # API 요청 파라미터 설정
    params = {
        "language": "ko-KR",               # 언어: 한국어
        "region": "KR",                    # 지역: 한국 개봉 기준
        "sort_by": "primary_release_date.desc", # 정렬: 개봉일 내림차순
        "primary_release_date.gte": start_date, # 시작일
        "primary_release_date.lte": end_date,   # 종료일
        "include_adult": "false"           # 성인 영화 제외
    }

    print(f"--- 데이터 수집 기간: {start_date} ~ {end_date} ---")

    for page in range(1, max_pages + 1):
        params["page"] = page
        response = requests.get(BASE_URL, headers=headers, params=params)
        
        if response.status_code == 200:
            data = response.json()
            movies = data.get('results', [])
            
            if not movies:
                print(f"페이지 {page}: 데이터가 더 이상 없습니다.")
                break
                
            all_movies.extend(movies)
            print(f"페이지 {page} 완료 (누적 {len(all_movies)}건)")
            
            # API 과부하 방지를 위한 미세한 지연 (선택 사항)
            time.sleep(0.1) 
        else:
            print(f"오류 발생 ({response.status_code}): {response.text}")
            break
            
    # 결과가 없을 경우 빈 DataFrame 반환
    if not all_movies:
        return pd.DataFrame()
        
    # DataFrame 생성 및 컬럼 정리
    df = pd.DataFrame(all_movies)
    
    # 분석에 유용한 주요 컬럼만 선택
    cols = ['id', 'title', 'release_date', 'vote_average', 'vote_count', 'popularity', 'overview']
    df = df[cols]
    
    # 개봉일 기준 정렬 (최신순)
    df = df.sort_values(by='release_date', ascending=False).reset_index(drop=True)
    
    return df

```
