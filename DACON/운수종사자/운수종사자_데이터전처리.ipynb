```python

def merge_data(df_idx, df_data, Test='A'):
    """
    Test 그룹을 기준으로 df_train과 df_train_A를 병합
    
    Args:
        df_train: 전체 train 데이터프레임
        df_train_A: Test 그룹 A 데이터프레임
        Test: 병합할 Test 그룹 (기본값: 'A')
    
    Returns:
        merged_df: 병합된 데이터프레임
    """
    # Test 그룹 필터링
    df_idx_filtered = df_idx[df_idx['Test'] == Test]
    df_data_filtered = df_data[df_data['Test'] == Test]
    
    # 병합 수행
    merged_df = df_idx_filtered.merge(
        df_data_filtered,
        on=['PrimaryKey', 'Test', 'TestDate'],
        how='inner',
        suffixes=('_idx', '_data')
    )
    
    return merged_df

if PROCESS['원본데이타정리']:
    # 함수 사용 예시
    df_train_merged_A = merge_data(df_train, df_train_A, Test='A')
    df_train_merged_B = merge_data(df_train, df_train_B, Test='B')
    df_test_merged_A = merge_data(df_test, df_test_A, Test='A')
    df_test_merged_B = merge_data(df_test, df_test_B, Test='B')

    # 결과 확인
    print(f"df_train_A 크기: {df_train_A.shape}")
    print(f"df_train_merged_A 크기: {df_train_merged_A.shape}")
    print(f"df_train_B 크기: {df_train_B.shape}")
    print(f"df_train_merged_B 크기: {df_train_merged_B.shape}")

    print(f"df_test_A 크기: {df_test_A.shape}")
    print(f"df_test_merged_A 크기: {df_test_merged_A.shape}")
    print(f"df_test_B 크기: {df_test_B.shape}")
    print(f"df_test_merged_B 크기: {df_test_merged_B.shape}")

    df_train_merged_A.head_att(3)
    df_train_merged_B.head_att(3)
    df_test_merged_A.head_att(3)
    df_test_merged_B.head_att(3)
```

```python
# 쉼표 개수를 계산하여 새로운 열 추가
def add_comma_count_columns(df, columns, test='A'):
    """
    각 열의 쉼표 개수를 계산하여 새로운 열에 추가합니다.
    
    Args:
        df: pandas DataFrame
        columns: 쉼표 개수를 계산할 열 이름 리스트
    
    Returns:
        df: 쉼표 개수 열이 추가된 DataFrame
    """
    for col in columns:
        new_col_name = f'C{col}'  # 새로운 열 이름 생성
        df[new_col_name] = df[col].astype(str).apply(lambda x: x.count(','))
    return df

columns_to_check_A = [
    'A1-1', 'A1-2', 'A1-3', 'A1-4', 'A2-1', 'A2-2', 'A2-3', 'A2-4',
    'A3-1', 'A3-2', 'A3-3', 'A3-4', 'A3-5', 'A3-6', 'A3-7',
    'A4-1', 'A4-2', 'A4-3', 'A4-4', 'A4-5',
    'A5-1', 'A5-2', 'A5-3', 'A6-1', 'A7-1', 'A8-1', 'A8-2',
    'A9-1', 'A9-2', 'A9-3', 'A9-4', 'A9-5',
]

columns_to_check_B = [
    'B1-1', 'B1-2', 'B1-3', 'B2-1', 'B2-2', 'B2-3',
    'B3-1', 'B3-2', 'B4-1', 'B4-2', 'B5-1', 'B5-2',
    'B6', 'B7', 'B8', 'B9-1', 'B9-2', 'B9-3', 'B9-4', 'B9-5',
    'B10-1', 'B10-2', 'B10-3', 'B10-4', 'B10-5', 'B10-6'
]

```

```python
def calc_A1(row):
    """
    A1 동적 시야 테스트 지수 계산
    
    입력:
        A1-1: direction (1=우→좌, 2=좌→우) - 18개
        A1-2: speed (1,2,3) - 18개
        A1-3: response_flag (0=응답안함, 1=응답함) - 18개
        A1-4: error (픽셀 오차) - 18개
        
    출력: 9개 지수
        - 우→좌(L): 실패율, 평균오차, 빠른속도실패, 표준편차 (4개)
        - 좌→우(R): 실패율, 평균오차, 빠른속도실패, 표준편차 (4개)
        - 방향 차이 (1개)
    """
    # 문자열을 numpy 배열로 변환
    direction = np.array([int(x) for x in str(row['A1-1']).split(',')])
    speed = np.array([int(x) for x in str(row['A1-2']).split(',')])
    response_flag = np.array([int(x) for x in str(row['A1-3']).split(',')])
    error = np.array([int(x) for x in str(row['A1-4']).split(',')])
    
    # 우→좌 (direction == 1)
    left_mask = (direction == 1)
    left_response = response_flag[left_mask]
    left_error = error[left_mask]
    
    A1L_fail_rate = (left_response == 0).sum() / len(left_response) if len(left_response) > 0 else 0.0
    
    # 응답한 경우만 오차 계산
    left_valid_error = np.abs(left_error[left_response == 1])
    A1L_mean_error = left_valid_error.mean() if len(left_valid_error) > 0 else 0.0
    A1L_std = left_valid_error.std() if len(left_valid_error) > 0 else 0.0
    
    # 빠른 속도(3)에서의 실패율
    left_fast_mask = left_mask & (speed == 3)
    left_fast_response = response_flag[left_fast_mask]
    A1L_fast_fail = (left_fast_response == 0).sum() / len(left_fast_response) if len(left_fast_response) > 0 else 0.0
    
    # 좌→우 (direction == 2)
    right_mask = (direction == 2)
    right_response = response_flag[right_mask]
    right_error = error[right_mask]
    
    A1R_fail_rate = (right_response == 0).sum() / len(right_response) if len(right_response) > 0 else 0.0
    
    right_valid_error = np.abs(right_error[right_response == 1])
    A1R_mean_error = right_valid_error.mean() if len(right_valid_error) > 0 else 0.0
    A1R_std = right_valid_error.std() if len(right_valid_error) > 0 else 0.0
    
    right_fast_mask = right_mask & (speed == 3)
    right_fast_response = response_flag[right_fast_mask]
    A1R_fast_fail = (right_fast_response == 0).sum() / len(right_fast_response) if len(right_fast_response) > 0 else 0.0
    
    # 방향 차이
    A1_direction_diff = abs(A1L_fail_rate - A1R_fail_rate)
    
    return pd.Series({
        'A1L_fail_rate': A1L_fail_rate,
        'A1L_mean_error': A1L_mean_error,
        'A1L_fast_fail': A1L_fast_fail,
        'A1L_std': A1L_std,
        'A1R_fail_rate': A1R_fail_rate,
        'A1R_mean_error': A1R_mean_error,
        'A1R_fast_fail': A1R_fast_fail,
        'A1R_std': A1R_std,
        'A1_direction_diff': A1_direction_diff
    })

logger.info("A1 동적 시야 테스트 변환 함수 생성 완료")
```

```python
def calc_A2(row):
    """
    A2 속도 변화 감지 테스트 지수 계산
    
    입력:
        A2-1: speed1 (첫번째 속도) - 18개
        A2-2: speed2 (두번째 속도) - 18개
        A2-3: response_flag (0=응답안함, 1=응답함) - 18개
        A2-4: error (픽셀 오차) - 18개
        
    출력: 9개 지수
        - 등속: 실패율, 평균오차 (2개)
        - 가속: 실패율, 평균오차 (2개)
        - 감속: 실패율, 평균오차 (2개)
        - 전체: 실패율, 평균오차, 표준편차 (3개)
    """
    # 문자열을 numpy 배열로 변환
    speed1 = np.array([int(x) for x in str(row['A2-1']).split(',')])
    speed2 = np.array([int(x) for x in str(row['A2-2']).split(',')])
    response_flag = np.array([int(x) for x in str(row['A2-3']).split(',')])
    error = np.array([int(x) for x in str(row['A2-4']).split(',')])
    
    # 속도 변화 계산
    speed_change = speed2 - speed1
    
    # 등속 (speed_change == 0)
    constant_mask = (speed_change == 0)
    constant_response = response_flag[constant_mask]
    A2_constant_fail = (constant_response == 0).sum() / len(constant_response) if len(constant_response) > 0 else 0.0
    
    constant_error = error[constant_mask]
    constant_valid_error = np.abs(constant_error[constant_response == 1])
    A2_constant_error = constant_valid_error.mean() if len(constant_valid_error) > 0 else 0.0
    
    # 가속 (speed_change > 0)
    accel_mask = (speed_change > 0)
    accel_response = response_flag[accel_mask]
    A2_accel_fail = (accel_response == 0).sum() / len(accel_response) if len(accel_response) > 0 else 0.0
    
    accel_error = error[accel_mask]
    accel_valid_error = np.abs(accel_error[accel_response == 1])
    A2_accel_error = accel_valid_error.mean() if len(accel_valid_error) > 0 else 0.0
    
    # 감속 (speed_change < 0)
    decel_mask = (speed_change < 0)
    decel_response = response_flag[decel_mask]
    A2_decel_fail = (decel_response == 0).sum() / len(decel_response) if len(decel_response) > 0 else 0.0
    
    decel_error = error[decel_mask]
    decel_valid_error = np.abs(decel_error[decel_response == 1])
    A2_decel_error = decel_valid_error.mean() if len(decel_valid_error) > 0 else 0.0
    
    # 전체
    A2_total_fail = (response_flag == 0).sum() / len(response_flag)
    
    valid_error = np.abs(error[response_flag == 1])
    A2_mean_error = valid_error.mean() if len(valid_error) > 0 else 0.0
    A2_std = valid_error.std() if len(valid_error) > 0 else 0.0
    
    return pd.Series({
        'A2_constant_fail': A2_constant_fail,
        'A2_constant_error': A2_constant_error,
        'A2_accel_fail': A2_accel_fail,
        'A2_accel_error': A2_accel_error,
        'A2_decel_fail': A2_decel_fail,
        'A2_decel_error': A2_decel_error,
        'A2_total_fail': A2_total_fail,
        'A2_mean_error': A2_mean_error,
        'A2_std': A2_std
    })

logger.info("A2 속도 변화 감지 테스트 변환 함수 생성 완료")
```

```python
def calc_A3(row):
    """
    A3 크기 비교 테스트 지수 계산
    
    입력:
        A3-1: size (1=Small, 2=Big) - 32개
        A3-2: stimulus1 - 32개
        A3-3: stimulus2_type (1=같음, 2=다름) - 32개
        A3-4: stimulus2 - 32개
        A3-5: response_type (1,2=valid, 3,4=invalid) - 32개
        A3-6: response_flag (0=응답안함, 1=응답함) - 32개
        A3-7: reaction_time (ms) - 32개
        
    출력: 12개 지수
        - Small: 정확도, 실패율, 평균RT (3개)
        - Big: 정확도, 실패율, 평균RT (3개)
        - Valid: 정확도, 실패율 (2개)
        - Invalid: 정확도, 실패율 (2개)
        - 크기차이, Valid/Invalid차이 (2개)
    """
    # 문자열을 numpy 배열로 변환
    size = np.array([int(x) for x in str(row['A3-1']).split(',')])
    response_type = np.array([int(x) for x in str(row['A3-5']).split(',')])
    response_flag = np.array([int(x) for x in str(row['A3-6']).split(',')])
    rt = np.array([int(x) for x in str(row['A3-7']).split(',')])
    
    # 정답 판단: response_type이 1 또는 3이면 정답
    correct = np.isin(response_type, [1, 3]).astype(int)
    
    # Small (size == 1)
    small_mask = (size == 1)
    small_correct = correct[small_mask]
    A3S_correct_rate = small_correct.mean() if len(small_correct) > 0 else 0.0
    
    small_response = response_flag[small_mask]
    A3S_fail_rate = (small_response == 0).sum() / len(small_response) if len(small_response) > 0 else 0.0
    
    small_rt = rt[small_mask & (response_flag == 1)]
    A3S_mean_rt = small_rt.mean() if len(small_rt) > 0 else 0.0
    
    # Big (size == 2)
    big_mask = (size == 2)
    big_correct = correct[big_mask]
    A3B_correct_rate = big_correct.mean() if len(big_correct) > 0 else 0.0
    
    big_response = response_flag[big_mask]
    A3B_fail_rate = (big_response == 0).sum() / len(big_response) if len(big_response) > 0 else 0.0
    
    big_rt = rt[big_mask & (response_flag == 1)]
    A3B_mean_rt = big_rt.mean() if len(big_rt) > 0 else 0.0
    
    # Valid (response_type in [1, 2])
    valid_mask = np.isin(response_type, [1, 2])
    valid_correct = correct[valid_mask]
    A3_valid_correct = valid_correct.mean() if len(valid_correct) > 0 else 0.0
    
    valid_response = response_flag[valid_mask]
    A3_valid_fail_rate = (valid_response == 0).sum() / len(valid_response) if len(valid_response) > 0 else 0.0
    
    # Invalid (response_type in [3, 4])
    invalid_mask = np.isin(response_type, [3, 4])
    invalid_correct = correct[invalid_mask]
    A3_invalid_correct = invalid_correct.mean() if len(invalid_correct) > 0 else 0.0
    
    invalid_response = response_flag[invalid_mask]
    A3_invalid_fail_rate = (invalid_response == 0).sum() / len(invalid_response) if len(invalid_response) > 0 else 0.0
    
    # 차이
    A3_size_diff = A3S_correct_rate - A3B_correct_rate
    A3_valid_invalid_diff = A3_valid_correct - A3_invalid_correct
    
    # 전체
    A3_total_fail = (response_flag == 0).sum() / len(response_flag)
    rt_valid = rt[response_flag == 1]
    A3_mean_rt = rt_valid.mean() if len(rt_valid) > 0 else 0.0
    
    return pd.Series({
        'A3S_correct_rate': A3S_correct_rate,
        'A3S_fail_rate': A3S_fail_rate,
        'A3S_mean_rt': A3S_mean_rt,
        'A3B_correct_rate': A3B_correct_rate,
        'A3B_fail_rate': A3B_fail_rate,
        'A3B_mean_rt': A3B_mean_rt,
        'A3_valid_correct': A3_valid_correct,
        'A3_valid_fail_rate': A3_valid_fail_rate,
        'A3_invalid_correct': A3_invalid_correct,
        'A3_invalid_fail_rate': A3_invalid_fail_rate,
        'A3_size_diff': A3_size_diff,
        'A3_valid_invalid_diff': A3_valid_invalid_diff,
        'A3_total_fail': A3_total_fail,
        'A3_mean_rt': A3_mean_rt
    })

logger.info("A3 크기 비교 테스트 변환 함수 생성 완료")
```

```python
def calc_A4(row):
    """
    A4 Stroop 테스트 지수 계산
    
    입력:
        A4-1: condition (1=Congruent, 2=Incongruent) - 80개
        A4-2: response - 80개
        A4-3: correct (1=정답, 2=오답) - 80개
        A4-4: response_flag (0=응답안함, 1=응답함) - 80개
        A4-5: reaction_time (ms) - 80개
        
    출력: 9개 지수
        - Congruent: 정확도, 실패율, 평균RT (3개)
        - Incongruent: 정확도, 실패율, 평균RT (3개)
        - Stroop 효과, 전체 정확도, 전체 실패율 (3개)
    """
    # 문자열을 numpy 배열로 변환
    condition = np.array([int(x) for x in str(row['A4-1']).split(',')])
    correct = np.array([int(x) for x in str(row['A4-3']).split(',')])
    response_flag = np.array([int(x) for x in str(row['A4-4']).split(',')])
    rt = np.array([int(x) for x in str(row['A4-5']).split(',')])
    
    # Congruent (condition == 1)
    con_mask = (condition == 1)
    con_correct = correct[con_mask]
    con_response = response_flag[con_mask]
    con_rt = rt[con_mask]
    
    # Congruent 지수
    A4C_correct_rate = (con_correct == 1).sum() / len(con_correct) if len(con_correct) > 0 else 0.0
    A4C_fail_rate = (con_response == 0).sum() / len(con_response) if len(con_response) > 0 else 0.0
    # 응답한 경우만 RT 계산
    con_rt_valid = con_rt[con_response == 1]
    A4C_mean_rt = con_rt_valid.mean() if len(con_rt_valid) > 0 else 0.0
    
    # Incongruent (condition == 2)
    incon_mask = (condition == 2)
    incon_correct = correct[incon_mask]
    incon_response = response_flag[incon_mask]
    incon_rt = rt[incon_mask]
    
    # Incongruent 지수
    A4I_correct_rate = (incon_correct == 1).sum() / len(incon_correct) if len(incon_correct) > 0 else 0.0
    A4I_fail_rate = (incon_response == 0).sum() / len(incon_response) if len(incon_response) > 0 else 0.0
    incon_rt_valid = incon_rt[incon_response == 1]
    A4I_mean_rt = incon_rt_valid.mean() if len(incon_rt_valid) > 0 else 0.0
    
    # Stroop 효과 (Congruent - Incongruent)
    # 정상: Con > Incon (양수), 비정상: Con < Incon (음수)
    A4_stroop_effect = A4C_correct_rate - A4I_correct_rate
    
    # 전체 지수
    A4_total_correct = (correct == 1).sum() / len(correct)
    A4_total_fail = (response_flag == 0).sum() / len(response_flag)
    rt_valid = rt[response_flag == 1]
    A4_mean_rt = rt_valid.mean() if len(rt_valid) > 0 else 0.0
    
    return pd.Series({
        'A4C_correct_rate': A4C_correct_rate,
        'A4C_fail_rate': A4C_fail_rate,
        'A4C_mean_rt': A4C_mean_rt,
        'A4I_correct_rate': A4I_correct_rate,
        'A4I_fail_rate': A4I_fail_rate,
        'A4I_mean_rt': A4I_mean_rt,
        'A4_stroop_effect': A4_stroop_effect,
        'A4_total_correct': A4_total_correct,
        'A4_total_fail': A4_total_fail,
        'A4_mean_rt': A4_mean_rt
    })

logger.info("A4 Stroop 테스트 변환 함수 생성 완료")
```

```python
def calc_A5(row):
    """
    A5 변화 감지 테스트 지수 계산
    
    입력:
        A5-1: condition (1=변화없음, 2,3,4=변화있음) - 36개
        A5-2: correct (1=정답, 2=오답) - 36개
        A5-3: response_flag (0=응답안함, 1=응답함) - 36개
        
    출력: 4개 지수
        - 변화없음 정확도
        - 변화있음 정확도
        - Sensitivity (변화감지능력)
        - 전체 실패율
    """
    # 문자열을 numpy 배열로 변환
    condition = np.array([int(x) for x in str(row['A5-1']).split(',')])
    correct = np.array([int(x) for x in str(row['A5-2']).split(',')])
    response_flag = np.array([int(x) for x in str(row['A5-3']).split(',')])
    
    # No Change (condition == 1)
    no_change_mask = (condition == 1)
    no_change_correct = correct[no_change_mask]
    A5_no_change_correct = (no_change_correct == 1).sum() / len(no_change_correct) if len(no_change_correct) > 0 else 0.0
    
    # Change (condition > 1)
    change_mask = (condition > 1)
    change_correct = correct[change_mask]
    A5_change_correct = (change_correct == 1).sum() / len(change_correct) if len(change_correct) > 0 else 0.0
    
    # Sensitivity (변화 감지 능력)
    # 정상: 변화있음 정확도가 높아야 함
    # 위험: 변화있음 정확도가 낮음 (음수가 큼)
    A5_sensitivity = A5_change_correct - A5_no_change_correct
    
    # 전체 실패율
    A5_total_fail = (response_flag == 0).sum() / len(response_flag)
    
    return pd.Series({
        'A5_no_change_correct': A5_no_change_correct,
        'A5_change_correct': A5_change_correct,
        'A5_sensitivity': A5_sensitivity,
        'A5_total_fail': A5_total_fail
    })

logger.info("A5 변화 감지 테스트 변환 함수 생성 완료")
```

```python
def calc_A6(row):
    """
    A6 문제풀이식 (판단능력) 테스트 지수 계산
    
    입력:
        A6-1: 정답 수 (총 14 trials 중) - 1개 값
        
    출력: 4개 지수
        - 정답 수 (1개)
        - 정답률 (1개)
        - 오답 수 (1개)
        - 정답률 구간 (1개): 1=낮음(0-50%), 2=중간(50-85%), 3=높음(85-100%)
    """
    # 정답 수 (단일 값)
    A6_correct_count = int(row['A6-1'])
    
    # 전체 시행 수
    total_trials = 14
    
    # 정답률 계산
    A6_accuracy = A6_correct_count / total_trials
    
    # 오답 수
    A6_incorrect_count = total_trials - A6_correct_count
    
    # 정답률 구간 분류
    if A6_accuracy >= 0.85:
        A6_acc_level = 3  # 높음 (85% 이상)
    elif A6_accuracy >= 0.50:
        A6_acc_level = 2  # 중간 (50-85%)
    else:
        A6_acc_level = 1  # 낮음 (50% 미만)
    
    return pd.Series({
        'A6_correct_count': A6_correct_count,
        'A6_accuracy': A6_accuracy,
        'A6_incorrect_count': A6_incorrect_count,
        'A6_acc_level': A6_acc_level
    })

logger.info("A6 문제풀이식 (판단능력) 테스트 변환 함수 완료")
```

```python
def calc_A7(row):
    """
    A7 문제풀이식 (단순도형) 테스트 지수 계산
    
    입력:
        A7-1: 정답 수 (총 18 trials 중) - 1개 값
        
    출력: 5개 지수
        - 정답 수 (1개)
        - 정답률 (1개)
        - 오답 수 (1개)
        - 정답률 구간 (1개): 1=위험(0-50%), 2=주의(50-75%), 3=보통(75-90%), 4=우수(90-100%)
        - 성능 레벨 (1개): 문자열 평가
    """
    # 정답 수 (단일 값)
    A7_correct_count = int(row['A7-1'])
    
    # 전체 시행 수
    total_trials = 18
    
    # 정답률 계산
    A7_accuracy = A7_correct_count / total_trials
    
    # 오답 수
    A7_incorrect_count = total_trials - A7_correct_count
    
    # 정답률 구간 분류
    if A7_accuracy >= 0.90:
        A7_acc_level = 4        # 우수 (90% 이상, 16개 이상)
        A7_performance = '우수'
    elif A7_accuracy >= 0.75:
        A7_acc_level = 3        # 보통 (75-90%, 14-15개)
        A7_performance = '보통'
    elif A7_accuracy >= 0.50:
        A7_acc_level = 2        # 주의 (50-75%, 9-13개)
        A7_performance = '주의'
    else:
        A7_acc_level = 1        # 위험 (50% 미만, 8개 이하)
        A7_performance = '위험'
    
    return pd.Series({
        'A7_correct_count': A7_correct_count,
        'A7_accuracy': A7_accuracy,
        'A7_incorrect_count': A7_incorrect_count,
        'A7_acc_level': A7_acc_level,
        'A7_performance': A7_performance
    })

logger.info("A7 문제풀이식 (단순도형) 테스트 변환 함수 생성 완료")
```

```python
def calc_A8(row):
    """
    A8 질문지형 (타당도 척도) 테스트 지수 계산
    
    입력:
        A8-1: 타당도 척도 1 (응답 왜곡 정도, 0~10) - 1개 값
        A8-2: 타당도 척도 2 (반응 일관성 정도, 0~10) - 1개 값
        
    출력: 8개 지수
        - 원점수 2개
        - 정규화 점수 2개
        - 이진 범주화 2개
        - 종합 신뢰도 2개
    """
    # 원점수 (단일 값)
    A8_distortion_score = int(row['A8-1'])      # 응답 왜곡
    A8_consistency_score = int(row['A8-2'])     # 반응 일관성
    
    # 정규화 (0~10 → 0~1)
    A8_distortion_norm = A8_distortion_score / 10.0
    A8_consistency_norm = A8_consistency_score / 10.0
    
    # 이진 범주화
    A8_distortion_high = 1 if A8_distortion_score >= 7 else 0      # 왜곡 높음 (≥7)
    A8_consistency_good = 1 if A8_consistency_score <= 2 else 0    # 일관성 좋음 (≤2)
    
    # 종합 타당도 점수 계산
    # 공식: 1 - (왜곡*0.1 + 일관성문제*0.15)
    # 왜곡 높을수록, 일관성 나쁠수록(높을수록) 점수 감소
    A8_validity_score = 1.0 - (A8_distortion_norm + A8_consistency_norm * 1.5) / 2.5
    A8_validity_score = max(0.0, min(1.0, A8_validity_score))  # 0~1 범위 클리핑
    
    # 검사 신뢰 가능 여부 (임계값: 0.7)
    A8_test_reliable = 1 if A8_validity_score >= 0.7 else 0
    
    return pd.Series({
        'A8_distortion_score': A8_distortion_score,
        'A8_consistency_score': A8_consistency_score,
        'A8_distortion_norm': A8_distortion_norm,
        'A8_consistency_norm': A8_consistency_norm,
        'A8_distortion_high': A8_distortion_high,
        'A8_consistency_good': A8_consistency_good,
        'A8_validity_score': A8_validity_score,
        'A8_test_reliable': A8_test_reliable
    })

logger.info("A8 질문지형 (타당도 척도) 테스트 변환 함수 생성 완료")
```

```python
def calc_A9(row):
    """
    A9 억제 능력 테스트 지수 계산 (Go/No-Go)
    
    입력:
        A9-1: condition (1=Go, 2=No-Go, 3=기타) - 48개
        A9-2: response (1=반응함, 2=반응안함) - 48개
        A9-3: correct (1=정답, 0=오답) - 48개
        A9-4: reaction_time (ms, 0=응답안함) - 48개
        A9-5: response_flag (0=응답안함, 1=응답함) - 48개
        
    출력: 28개 지수
        - 반응시간 통계 7개
        - 전체 정확도 1개
        - Go 시행 3개 (RT, 정확도, 놓침)
        - No-Go 시행 3개 (정확도, 억제실패, RT)
        - 조건별 통계 6개
        - 오답 패턴 2개
        - 이상 반응 2개
        - 응답 패턴 4개
    """
    # 문자열을 numpy 배열로 변환
    condition = np.array([int(x) for x in str(row['A9-1']).split(',')])
    response = np.array([int(x) for x in str(row['A9-2']).split(',')])
    correct = np.array([int(x) for x in str(row['A9-3']).split(',')])
    rt = np.array([int(x) for x in str(row['A9-4']).split(',')])
    response_flag = np.array([int(x) for x in str(row['A9-5']).split(',')])
    
    # 반응시간 통계 (응답한 경우만)
    rt_valid = rt[rt > 0]
    if len(rt_valid) > 0:
        A9_rt_mean = rt_valid.mean()
        A9_rt_std = rt_valid.std()
        A9_rt_cv = A9_rt_std / A9_rt_mean if A9_rt_mean > 0 else 0.0
        A9_rt_min = rt_valid.min()
        A9_rt_max = rt_valid.max()
        A9_rt_q25 = np.percentile(rt_valid, 25)
        A9_rt_q75 = np.percentile(rt_valid, 75)
    else:
        A9_rt_mean = A9_rt_std = A9_rt_cv = A9_rt_min = A9_rt_max = A9_rt_q25 = A9_rt_q75 = 0.0
    
    # 전체 정확도
    A9_accuracy = (correct == 1).sum() / len(correct)
    
    # Go 시행 (condition == 1)
    go_mask = (condition == 1)
    if go_mask.sum() > 0:
        go_rt = rt[go_mask & (rt > 0)]
        A9_go_rt = go_rt.mean() if len(go_rt) > 0 else 0.0
        
        go_correct = correct[go_mask]
        A9_go_acc = (go_correct == 1).sum() / len(go_correct)
        
        # Go 놓침 (응답 안한 경우)
        go_response = response_flag[go_mask]
        A9_go_miss = (go_response == 0).sum() / len(go_response)
    else:
        A9_go_rt = A9_go_acc = A9_go_miss = 0.0
    
    # No-Go 시행 (condition == 2)
    nogo_mask = (condition == 2)
    if nogo_mask.sum() > 0:
        nogo_correct = correct[nogo_mask]
        # 정답: 반응 안함 (억제 성공)
        A9_nogo_acc = (nogo_correct == 1).sum() / len(nogo_correct)
        
        # 억제 실패 (반응한 경우)
        A9_nogo_commission = (nogo_correct == 0).sum() / len(nogo_correct)
        
        # No-Go에서 잘못 반응한 경우의 RT
        nogo_rt = rt[nogo_mask & (rt > 0)]
        A9_nogo_rt = nogo_rt.mean() if len(nogo_rt) > 0 else 0.0
    else:
        A9_nogo_acc = A9_nogo_commission = A9_nogo_rt = 0.0
    
    # 조건별 통계
    # Condition 1 (Go)
    if go_mask.sum() > 0:
        cond1_rt = rt[go_mask & (rt > 0)]
        A9_cond1_rt = cond1_rt.mean() if len(cond1_rt) > 0 else 0.0
        cond1_correct = correct[go_mask]
        A9_cond1_acc = (cond1_correct == 1).sum() / len(cond1_correct)
    else:
        A9_cond1_rt = A9_cond1_acc = 0.0
    
    # Condition 2 (No-Go)
    if nogo_mask.sum() > 0:
        cond2_rt = rt[nogo_mask & (rt > 0)]
        A9_cond2_rt = cond2_rt.mean() if len(cond2_rt) > 0 else 0.0
        cond2_correct = correct[nogo_mask]
        A9_cond2_acc = (cond2_correct == 1).sum() / len(cond2_correct)
    else:
        A9_cond2_rt = A9_cond2_acc = 0.0
    
    # Condition 3 (기타)
    cond3_mask = (condition == 3)
    if cond3_mask.sum() > 0:
        cond3_rt = rt[cond3_mask & (rt > 0)]
        A9_cond3_rt = cond3_rt.mean() if len(cond3_rt) > 0 else 0.0
        cond3_correct = correct[cond3_mask]
        A9_cond3_acc = (cond3_correct == 1).sum() / len(cond3_correct)
    else:
        A9_cond3_rt = A9_cond3_acc = 0.0
    
    # 오답 패턴
    error_rt = rt[(correct == 0) & (rt > 0)]
    A9_error_rt = error_rt.mean() if len(error_rt) > 0 else 0.0
    
    correct_rt = rt[(correct == 1) & (rt > 0)]
    A9_correct_rt = correct_rt.mean() if len(correct_rt) > 0 else 0.0
    
    # 이상 반응
    A9_too_fast_cnt = (rt_valid < 300).sum() if len(rt_valid) > 0 else 0
    A9_too_slow_cnt = (rt_valid > 600).sum() if len(rt_valid) > 0 else 0
    
    # 응답 패턴
    A9_response_rate = (response_flag == 1).sum() / len(response_flag)
    A9_no_response_cnt = (response_flag == 0).sum()
    
    # False Alarm (No-Go에서 반응함)
    A9_false_alarm_rate = A9_nogo_commission
    
    # 충동성 지수 (No-Go 잘못 반응 수 / 전체)
    if nogo_mask.sum() > 0:
        nogo_error_cnt = (nogo_correct == 0).sum()
        A9_impulsivity_index = nogo_error_cnt / len(correct)
    else:
        A9_impulsivity_index = 0.0
    
    return pd.Series({
        'A9_rt_mean': A9_rt_mean,
        'A9_rt_std': A9_rt_std,
        'A9_rt_cv': A9_rt_cv,
        'A9_rt_min': A9_rt_min,
        'A9_rt_max': A9_rt_max,
        'A9_rt_q25': A9_rt_q25,
        'A9_rt_q75': A9_rt_q75,
        'A9_accuracy': A9_accuracy,
        'A9_go_rt': A9_go_rt,
        'A9_go_acc': A9_go_acc,
        'A9_go_miss': A9_go_miss,
        'A9_nogo_acc': A9_nogo_acc,
        'A9_nogo_commission': A9_nogo_commission,
        'A9_nogo_rt': A9_nogo_rt,
        'A9_cond1_rt': A9_cond1_rt,
        'A9_cond1_acc': A9_cond1_acc,
        'A9_cond2_rt': A9_cond2_rt,
        'A9_cond2_acc': A9_cond2_acc,
        'A9_cond3_rt': A9_cond3_rt,
        'A9_cond3_acc': A9_cond3_acc,
        'A9_error_rt': A9_error_rt,
        'A9_correct_rt': A9_correct_rt,
        'A9_too_fast_cnt': A9_too_fast_cnt,
        'A9_too_slow_cnt': A9_too_slow_cnt,
        'A9_response_rate': A9_response_rate,
        'A9_no_response_cnt': A9_no_response_cnt,
        'A9_false_alarm_rate': A9_false_alarm_rate,
        'A9_impulsivity_index': A9_impulsivity_index
    })

logger.info("A9 억제 능력 테스트 변환 함수 생성 완료")
```

```python
def calc_B1(row):
    """
    B1 선택 반응시간 테스트 지수 계산
    
    입력:
        B1-1: 자극조건 (36개) - 1,2
        B1-2: 응답값 (36개) - 0~1 범위, 0=무응답
        B1-3: 정답여부 (36개) - 1=정답, 4=오답/무응답
        
    출력: 20개 지수
    """
    try:
        # 문자열을 배열로 변환
        stimulus = np.array([int(x) for x in str(row['B1-1']).split(',')])
        response = np.array([float(x) for x in str(row['B1-2']).split(',')])
        correct_code = np.array([int(x) for x in str(row['B1-3']).split(',')])
        
        # 정답 판단: 1=정답, 4=오답
        correct = (correct_code == 1).astype(int)
        
        # 반응한 경우만 (response > 0)
        valid_responses = response[response > 0]
        
        # 반응시간 통계 (7개)
        B1_rt_mean = valid_responses.mean() if len(valid_responses) > 0 else 0.0
        B1_rt_std = valid_responses.std() if len(valid_responses) > 0 else 0.0
        B1_rt_cv = B1_rt_std / B1_rt_mean if B1_rt_mean > 0 else 0.0
        B1_rt_min = valid_responses.min() if len(valid_responses) > 0 else 0.0
        B1_rt_max = valid_responses.max() if len(valid_responses) > 0 else 0.0
        B1_rt_q25 = np.percentile(valid_responses, 25) if len(valid_responses) > 0 else 0.0
        B1_rt_q75 = np.percentile(valid_responses, 75) if len(valid_responses) > 0 else 0.0
        
        # 정확도 (1개)
        B1_accuracy = correct.mean()
        
        # 자극별 반응시간 (3개)
        stim1_mask = (stimulus == 1) & (response > 0)
        stim2_mask = (stimulus == 2) & (response > 0)
        B1_stim1_rt = response[stim1_mask].mean() if stim1_mask.sum() > 0 else 0.0
        B1_stim2_rt = response[stim2_mask].mean() if stim2_mask.sum() > 0 else 0.0
        B1_stim3_rt = 0.0  # 조건3 없음
        
        # 자극별 정확도 (3개)
        B1_stim1_acc = correct[stimulus == 1].mean() if (stimulus == 1).sum() > 0 else 0.0
        B1_stim2_acc = correct[stimulus == 2].mean() if (stimulus == 2).sum() > 0 else 0.0
        B1_stim3_acc = 0.0
        
        # 구간별 반응시간 (3개) - 12개씩
        n = len(response)
        third = n // 3
        early_mask = (response[:third] > 0)
        middle_mask = (response[third:2*third] > 0)
        late_mask = (response[2*third:] > 0)
        
        B1_early_rt = response[:third][early_mask].mean() if early_mask.sum() > 0 else 0.0
        B1_middle_rt = response[third:2*third][middle_mask].mean() if middle_mask.sum() > 0 else 0.0
        B1_late_rt = response[2*third:][late_mask].mean() if late_mask.sum() > 0 else 0.0
        
        # 피로도 (1개)
        B1_fatigue_index = B1_late_rt - B1_early_rt
        
        # 응답 패턴 (4개)
        B1_response_rate = (response > 0).sum() / len(response)
        B1_no_response_cnt = (response == 0).sum()
        B1_response_mean = valid_responses.mean() if len(valid_responses) > 0 else 0.0
        B1_response_consistency = 1 - B1_rt_cv
        
        # 이상 반응 (2개)
        B1_too_fast_cnt = (valid_responses < 0.5).sum()
        B1_too_slow_cnt = (valid_responses > 0.85).sum()
        
        return pd.Series({
            'B1_rt_mean': B1_rt_mean,
            'B1_rt_std': B1_rt_std,
            'B1_rt_cv': B1_rt_cv,
            'B1_rt_min': B1_rt_min,
            'B1_rt_max': B1_rt_max,
            'B1_rt_q25': B1_rt_q25,
            'B1_rt_q75': B1_rt_q75,
            'B1_accuracy': B1_accuracy,
            'B1_stim1_rt': B1_stim1_rt,
            'B1_stim2_rt': B1_stim2_rt,
            'B1_stim3_rt': B1_stim3_rt,
            'B1_stim1_acc': B1_stim1_acc,
            'B1_stim2_acc': B1_stim2_acc,
            'B1_stim3_acc': B1_stim3_acc,
            'B1_early_rt': B1_early_rt,
            'B1_middle_rt': B1_middle_rt,
            'B1_late_rt': B1_late_rt,
            'B1_fatigue_index': B1_fatigue_index,
            'B1_response_rate': B1_response_rate,
            'B1_no_response_cnt': B1_no_response_cnt,
            'B1_response_mean': B1_response_mean,
            'B1_response_consistency': B1_response_consistency,
            'B1_too_fast_cnt': B1_too_fast_cnt,
            'B1_too_slow_cnt': B1_too_slow_cnt
        })
    except Exception as e:
        logger.error(f"calc_B1 오류: {e}")
        return pd.Series({f'B1_{k}': 0.0 for k in ['rt_mean', 'rt_std', 'rt_cv', 'rt_min', 'rt_max', 
                                                     'rt_q25', 'rt_q75', 'accuracy', 'stim1_rt', 'stim2_rt',
                                                     'stim3_rt', 'stim1_acc', 'stim2_acc', 'stim3_acc',
                                                     'early_rt', 'middle_rt', 'late_rt', 'fatigue_index',
                                                     'response_rate', 'no_response_cnt', 'response_mean',
                                                     'response_consistency', 'too_fast_cnt', 'too_slow_cnt']})

logger.info("B1 선택 반응시간 테스트 변환 함수 생성 완료")
```

```python
def calc_B2(row):
    """
    B2 선택 반응시간 테스트 (확장) 지수 계산
    
    입력:
        B2-1: 자극조건 (24개)
        B2-2: 응답값 (24개) - 0~1 범위
        B2-3: 정답여부 (24개) - 1=정답, 3=특정오답, 4=일반오답/무응답
        
    출력: 22개 지수
    """
    try:
        stimulus = np.array([int(x) for x in str(row['B2-1']).split(',')])
        response = np.array([float(x) for x in str(row['B2-2']).split(',')])
        correct_code = np.array([int(x) for x in str(row['B2-3']).split(',')])
        
        correct = (correct_code == 1).astype(int)
        valid_responses = response[response > 0]
        
        # 반응시간 통계 (7개)
        B2_rt_mean = valid_responses.mean() if len(valid_responses) > 0 else 0.0
        B2_rt_std = valid_responses.std() if len(valid_responses) > 0 else 0.0
        B2_rt_cv = B2_rt_std / B2_rt_mean if B2_rt_mean > 0 else 0.0
        B2_rt_min = valid_responses.min() if len(valid_responses) > 0 else 0.0
        B2_rt_max = valid_responses.max() if len(valid_responses) > 0 else 0.0
        B2_rt_q25 = np.percentile(valid_responses, 25) if len(valid_responses) > 0 else 0.0
        B2_rt_q75 = np.percentile(valid_responses, 75) if len(valid_responses) > 0 else 0.0
        
        # 정확도 (1개)
        B2_accuracy = correct.mean()
        
        # 자극별 (3개 + 3개)
        stim1_mask = (stimulus == 1) & (response > 0)
        stim2_mask = (stimulus == 2) & (response > 0)
        B2_stim1_rt = response[stim1_mask].mean() if stim1_mask.sum() > 0 else 0.0
        B2_stim2_rt = response[stim2_mask].mean() if stim2_mask.sum() > 0 else 0.0
        B2_stim3_rt = 0.0
        
        B2_stim1_acc = correct[stimulus == 1].mean() if (stimulus == 1).sum() > 0 else 0.0
        B2_stim2_acc = correct[stimulus == 2].mean() if (stimulus == 2).sum() > 0 else 0.0
        B2_stim3_acc = 0.0
        
        # 구간별 (3개)
        n = len(response)
        third = n // 3
        early_mask = (response[:third] > 0)
        middle_mask = (response[third:2*third] > 0)
        late_mask = (response[2*third:] > 0)
        
        B2_early_rt = response[:third][early_mask].mean() if early_mask.sum() > 0 else 0.0
        B2_middle_rt = response[third:2*third][middle_mask].mean() if middle_mask.sum() > 0 else 0.0
        B2_late_rt = response[2*third:][late_mask].mean() if late_mask.sum() > 0 else 0.0
        
        # 피로도 (1개)
        B2_fatigue_index = B2_late_rt - B2_early_rt
        
        # 응답 패턴 (4개)
        B2_response_rate = (response > 0).sum() / len(response)
        B2_no_response_cnt = (response == 0).sum()
        B2_response_mean = valid_responses.mean() if len(valid_responses) > 0 else 0.0
        B2_response_consistency = 1 - B2_rt_cv
        
        # 오답 유형 분석 (3개)
        B2_type1_error_rate = (correct_code == 4).sum() / len(correct_code)
        B2_type3_error_rate = (correct_code == 3).sum() / len(correct_code)
        correct_cnt = (correct_code == 1).sum()
        correct_with_response = ((correct_code == 1) & (response > 0)).sum()
        B2_correct_response_rate = correct_with_response / correct_cnt if correct_cnt > 0 else 0.0
        
        # 이상 반응 (2개)
        B2_too_fast_cnt = (valid_responses < 0.4).sum()
        B2_too_slow_cnt = (valid_responses > 0.9).sum()
        
        return pd.Series({
            'B2_rt_mean': B2_rt_mean, 'B2_rt_std': B2_rt_std, 'B2_rt_cv': B2_rt_cv,
            'B2_rt_min': B2_rt_min, 'B2_rt_max': B2_rt_max, 'B2_rt_q25': B2_rt_q25, 'B2_rt_q75': B2_rt_q75,
            'B2_accuracy': B2_accuracy,
            'B2_stim1_rt': B2_stim1_rt, 'B2_stim2_rt': B2_stim2_rt, 'B2_stim3_rt': B2_stim3_rt,
            'B2_stim1_acc': B2_stim1_acc, 'B2_stim2_acc': B2_stim2_acc, 'B2_stim3_acc': B2_stim3_acc,
            'B2_early_rt': B2_early_rt, 'B2_middle_rt': B2_middle_rt, 'B2_late_rt': B2_late_rt,
            'B2_fatigue_index': B2_fatigue_index,
            'B2_response_rate': B2_response_rate, 'B2_no_response_cnt': B2_no_response_cnt,
            'B2_response_mean': B2_response_mean, 'B2_response_consistency': B2_response_consistency,
            'B2_type1_error_rate': B2_type1_error_rate, 'B2_type3_error_rate': B2_type3_error_rate,
            'B2_correct_response_rate': B2_correct_response_rate,
            'B2_too_fast_cnt': B2_too_fast_cnt, 'B2_too_slow_cnt': B2_too_slow_cnt
        })
    except Exception as e:
        logger.error(f"calc_B2 오류: {e}")
        return pd.Series({f'B2_{k}': 0.0 for k in ['rt_mean', 'rt_std', 'rt_cv', 'rt_min', 'rt_max',
                                                     'rt_q25', 'rt_q75', 'accuracy', 'stim1_rt', 'stim2_rt', 'stim3_rt',
                                                     'stim1_acc', 'stim2_acc', 'stim3_acc', 'early_rt', 'middle_rt', 'late_rt',
                                                     'fatigue_index', 'response_rate', 'no_response_cnt', 'response_mean',
                                                     'response_consistency', 'type1_error_rate', 'type3_error_rate',
                                                     'correct_response_rate', 'too_fast_cnt', 'too_slow_cnt']})

logger.info("B2 선택 반응시간 테스트 (확장) 변환 함수 생성 완료")
```

```python
def calc_B6(row):
    """
    B6 색상 변별 테스트 지수 계산
    
    입력:
        B6: 단일 값 (15개) - 데이터 불충분
        
    출력: 4개 지수 (최소)
    """
    try:
        values = np.array([int(x) for x in str(row['B6']).split(',')])
        
        # 최소한의 정보만 추출
        B6_sample_size = len(values)
        B6_value1_count = (values == 1).sum()
        B6_value2_count = (values == 2).sum()
        B6_value1_ratio = B6_value1_count / B6_sample_size if B6_sample_size > 0 else 0.0
        
        return pd.Series({
            'B6_sample_size': B6_sample_size,
            'B6_value1_count': B6_value1_count,
            'B6_value2_count': B6_value2_count,
            'B6_value1_ratio': B6_value1_ratio
        })
    except Exception as e:
        logger.error(f"calc_B6 오류: {e}")
        return pd.Series({'B6_sample_size': 0, 'B6_value1_count': 0, 
                         'B6_value2_count': 0, 'B6_value1_ratio': 0.0})

logger.info("B6 색상 변별 테스트 변환 함수 생성 완료 (데이터 불충분)")
```

```python
def calc_B7(row):
    """
    B7 주의 전환 테스트 지수 계산
    
    입력:
        B7: 단일 값 (15개) - 데이터 불충분
        
    출력: 8개 지수 (최소)
    """
    try:
        values = np.array([int(x) for x in str(row['B7']).split(',')])
        
        # 자극 분포로 해석
        B7_type1_count = (values == 1).sum()
        B7_type2_count = (values == 2).sum()
        B7_type1_ratio = B7_type1_count / len(values) if len(values) > 0 else 0.0
        B7_type2_ratio = B7_type2_count / len(values) if len(values) > 0 else 0.0
        
        # 전환 횟수
        switches = 0
        for i in range(1, len(values)):
            if values[i] != values[i-1]:
                switches += 1
        B7_switch_count = switches
        B7_switch_rate = switches / len(values) if len(values) > 0 else 0.0
        
        B7_sample_size = len(values)
        B7_data_completeness = len(values) / 24  # 정상 24개
        
        return pd.Series({
            'B7_type1_count': B7_type1_count,
            'B7_type2_count': B7_type2_count,
            'B7_type1_ratio': B7_type1_ratio,
            'B7_type2_ratio': B7_type2_ratio,
            'B7_switch_count': B7_switch_count,
            'B7_switch_rate': B7_switch_rate,
            'B7_sample_size': B7_sample_size,
            'B7_data_completeness': B7_data_completeness
        })
    except Exception as e:
        logger.error(f"calc_B7 오류: {e}")
        return pd.Series({f'B7_{k}': 0.0 for k in ['type1_count', 'type2_count', 'type1_ratio',
                                                     'type2_ratio', 'switch_count', 'switch_rate',
                                                     'sample_size', 'data_completeness']})

logger.info("B7 주의 전환 테스트 변환 함수 생성 완료 (데이터 불충분)")

```

```python
def calc_B8(row):
    """
    B8 작업기억 (3-back) 테스트 지수 계산
    
    입력:
        B8: 단일 값 (12개) - 모두 1, 데이터 불충분
        
    출력: 7개 지수 (최소)
    """
    try:
        values = np.array([int(x) for x in str(row['B8']).split(',')])
        
        # 기본 통계만
        B8_accuracy = 1.0  # 가정
        B8_error_count = 0
        B8_total_trials = len(values)
        B8_data_completeness = len(values) / 32  # 정상 32개
        B8_response_uniformity = 1.0 if len(set(values)) == 1 else 0.0
        B8_unique_values = len(set(values))
        B8_mean_value = values.mean() if len(values) > 0 else 0.0
        
        return pd.Series({
            'B8_accuracy': B8_accuracy,
            'B8_error_count': B8_error_count,
            'B8_total_trials': B8_total_trials,
            'B8_data_completeness': B8_data_completeness,
            'B8_response_uniformity': B8_response_uniformity,
            'B8_unique_values': B8_unique_values,
            'B8_mean_value': B8_mean_value
        })
    except Exception as e:
        logger.error(f"calc_B8 오류: {e}")
        return pd.Series({f'B8_{k}': 0.0 for k in ['accuracy', 'error_count', 'total_trials',
                                                     'data_completeness', 'response_uniformity',
                                                     'unique_values', 'mean_value']})

logger.info("B8 작업기억 (3-back) 테스트 변환 함수 생성 완료 (데이터 불충분)")

```

```python
def calc_B9(row):
    """
    B9 Go/No-Go 테스트 지수 계산
    
    입력:
        B9-1~B9-5: 각 단일 값 - 데이터 불충분
        
    출력: 6개 지수 (최소)
    """
    try:
        b9_1 = float(row['B9-1']) if pd.notna(row['B9-1']) else 0.0
        b9_2 = float(row['B9-2']) if pd.notna(row['B9-2']) else 0.0
        b9_3 = float(row['B9-3']) if pd.notna(row['B9-3']) else 0.0
        b9_4 = float(row['B9-4']) if pd.notna(row['B9-4']) else 0.0
        b9_5 = float(row['B9-5']) if pd.notna(row['B9-5']) else 0.0
        
        # 집계값으로 추정 해석
        B9_value_sum = b9_1 + b9_2 + b9_3 + b9_4 + b9_5
        B9_value_mean = B9_value_sum / 5
        B9_nonzero_count = sum([1 for v in [b9_1, b9_2, b9_3, b9_4, b9_5] if v > 0])
        B9_max_value = max([b9_1, b9_2, b9_3, b9_4, b9_5])
        B9_min_value = min([b9_1, b9_2, b9_3, b9_4, b9_5])
        B9_value_range = B9_max_value - B9_min_value
        
        return pd.Series({
            'B9_value_sum': B9_value_sum,
            'B9_value_mean': B9_value_mean,
            'B9_nonzero_count': B9_nonzero_count,
            'B9_max_value': B9_max_value,
            'B9_min_value': B9_min_value,
            'B9_value_range': B9_value_range
        })
    except Exception as e:
        logger.error(f"calc_B9 오류: {e}")
        return pd.Series({f'B9_{k}': 0.0 for k in ['value_sum', 'value_mean', 'nonzero_count',
                                                     'max_value', 'min_value', 'value_range']})

logger.info("B9 Go/No-Go 테스트 변환 함수 생성 완료 (데이터 불충분)")
```
