`safety_settings`는 Google GenAI API에서 생성된 콘텐츠의 안전성을 제어하기 위해 사용됩니다. 이 설정은 특정 카테고리의 유해 콘텐츠를 감지하고, 이를 차단하거나 제한하는 데 사용됩니다. 각 카테고리는 특정 유형의 유해성을 나타내며, `threshold`는 해당 카테고리에서 콘텐츠를 차단할 민감도를 설정합니다.

- 주요 구성 요소:
1. **`category`**: 유해 콘텐츠의 유형을 정의합니다. 예: 괴롭힘, 증오 발언 등.
2. **`threshold`**: 차단 민감도를 설정합니다. 가능한 값은 다음과 같습니다:
   - `BLOCK_NONE`: 차단하지 않음.
   - `BLOCK_LOW_AND_ABOVE`: 낮은 수준 이상의 유해 콘텐츠 차단.
   - `BLOCK_MEDIUM_AND_ABOVE`: 중간 수준 이상의 유해 콘텐츠 차단.
   - `BLOCK_ONLY_HIGH`: 높은 수준의 유해 콘텐츠만 차단.


- 사용 가능한 카테고리 및 `threshold` 값

| **카테고리**                     | **설명**                              | **가능한 Threshold 값**            |
|----------------------------------|---------------------------------------|------------------------------------|
| `HARM_CATEGORY_HARASSMENT`       | 괴롭힘 및 부적절한 상호작용           | `BLOCK_NONE`, `BLOCK_LOW_AND_ABOVE`, `BLOCK_MEDIUM_AND_ABOVE`, `BLOCK_ONLY_HIGH` |
| `HARM_CATEGORY_HATE_SPEECH`      | 증오 발언 및 차별적 표현               | `BLOCK_NONE`, `BLOCK_LOW_AND_ABOVE`, `BLOCK_MEDIUM_AND_ABOVE`, `BLOCK_ONLY_HIGH` |
| `HARM_CATEGORY_VIOLENCE`         | 폭력적인 콘텐츠                        | `BLOCK_NONE`, `BLOCK_LOW_AND_ABOVE`, `BLOCK_MEDIUM_AND_ABOVE`, `BLOCK_ONLY_HIGH` |
| `HARM_CATEGORY_SELF_HARM`        | 자해 또는 자살 관련 콘텐츠             | `BLOCK_NONE`, `BLOCK_LOW_AND_ABOVE`, `BLOCK_MEDIUM_AND_ABOVE`, `BLOCK_ONLY_HIGH` |
| `HARM_CATEGORY_ADULT`            | 성인용 콘텐츠                          | `BLOCK_NONE`, `BLOCK_LOW_AND_ABOVE`, `BLOCK_MEDIUM_AND_ABOVE`, `BLOCK_ONLY_HIGH` |
| `HARM_CATEGORY_DANGEROUS_GOODS`  | 위험한 물질 또는 활동 관련 콘텐츠      | `BLOCK_NONE`, `BLOCK_LOW_AND_ABOVE`, `BLOCK_MEDIUM_AND_ABOVE`, `BLOCK_ONLY_HIGH` |

---

`types.GenerateContentConfig`는 Google Gemini API를 사용하여 콘텐츠를 생성할 때, 생성 방식에 대한 다양한 설정을 담는 구성 클래스입니다. `generate_content`와 같은 함수의 `config` 매개변수에 이 클래스의 인스턴스를 전달하여 모델의 동작을 제어할 수 있습니다.

주요 매개변수는 다음과 같습니다.

*   **`temperature`**: (0.0 ~ 1.0) 생성될 텍스트의 무작위성을 조절합니다. 값이 높을수록 더 창의적이고 다양한 결과가 나오며, 낮을수록 더 예측 가능하고 일관된 결과가 나옵니다.
*   **`top_p`**: (0.0 ~ 1.0) Nucleus 샘플링 방식을 사용합니다. 모델이 다음 단어를 선택할 때, 확률의 합이 `top_p`가 될 때까지의 상위 토큰들만 고려합니다. `temperature`와 함께 무작위성을 조절하는 데 사용됩니다.
*   **`top_k`**: (정수) 모델이 다음 단어를 선택할 때 고려할 가장 확률이 높은 토큰의 개수를 제한합니다.
*   **`max_output_tokens`**: (정수) 생성될 응답의 최대 토큰(단어 또는 단어의 일부) 수를 제한합니다.
*   **`stop_sequences`**: (문자열 리스트) 모델이 특정 문자열을 생성하면 응답 생성을 중단하도록 설정합니다.
*   **`safety_settings`**: 유해 콘텐츠(예: 증오 발언, 폭력적인 내용)를 차단하는 기준을 설정합니다.

```python
# 대화 히스토리 초기화
chat_history = []

def print_chat_history(chat_history):
    for idx, content in enumerate(chat_history):
        logger.info(f"[{idx}] {'-' * 80}")
        for part in content.parts:
            logger.info(f"[{idx}] {part.text}")
            
def send_message(user_input):
    # 사용자 메시지 추가
    chat_history.append(
        types.Content(
            role='user',
            parts=[types.Part.from_text(text=user_input)]
        )
    )
    
    print_chat_history(chat_history)
    
    # 전체 히스토리와 함께 요청
    response = client.models.generate_content(
        model='gemini-2.0-flash-exp',
        contents=chat_history
    )
    
    # 모델 응답 추가
    chat_history.append(
        types.Content(
            role='model',
            parts=[types.Part.from_text(text=response.text)]
        )
    )
    
    logger.info('=' * 80)
        
    logger.info(f' >> {response.text}')
    return response
```
